```python
company_names = ['Apple', 'Microsoft', 'Tesla']
tickers = ['AAPL', 'MSFT', 'TSLA']
z = zip(company_names, tickers)
print(z)
```

```python
stocks = ["AAPL", "MSFT", "TSL"]
iterator = iter(stocks)
while(True):
    try:
        stock = next(iterator)
        print(stock)
    except StopIteration:
        break 
```
