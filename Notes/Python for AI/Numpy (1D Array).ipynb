# Extract or Select

**Accessing elements/ Extracting the values**

1. Accessing a single value through Indexing
2. Accessing a multiple values based on slicing (when there is pattern)
3. Accessing a multiple values based on indexing (when there is no pattern)
4. Accessing a single or multiple values based on condition

```python
a = np.array([15,19,24,32,37,42,46])
print(a)
```

**1. Accessing a single value through Indexing**

```python
#positive indexing
print(a[3])
```

```python
print(a[-4])
```

**2. Accessing a multiple values based on slicing**

```python
a[0:7:2]
```

```python
a[1:4]
```

```python
a[1:4:-1]
```

**3. Accessing a multiple values based on indexing**

Syntax : array[list of index numbers]

```python
a[[0,1,4,5]]
```

```python
a[[0,1,2,0,2,0,3]]
```

**4. Accessing a multiple values based on condition (Mask Indexing or Fancy Indexing)**

Syntax: array[condition]

-- first, it will check the condition, the item which satisifies the condition (means which item gives True for the condition), that item will be selected

```python
a[a>30]
```

```python
#inside a, select data where condition is True 
a[a%2==0]
```

```python
a[(a<30) | (a>40)]
```

# Data Manipulation or Data Modification in 1D Array

- Insert or Add
- Delete or Remove
- Update or Replace

```python
a = np.array([15,19,24])
print(a)
```

**Add a value in a array**4

syntax to add a single value : np.append(array,singlevalue)

syntax to add a multiple values : np.append(array,list of values)

```python
a1 = np.append(a,50)
print(a1)
```

```python
a2 = np.append(a,[60,70,80])
print(a2)
```

**Replacing a single index position with single value in a 1D array**

```python
a  = np.array([15,19,24,37,42,26])
print(a)
```

```python
a[1] = 27
print(a)
```

**Replacing a multiple index positions with a single value in a array**

```python
a[[0,1,4,5]] = 10
print(a)
```

**Replacing a multiple index positions with a multiple values in a array**

```python
a[[0,1,4]] = [20,19,45]
print(a)
```

**Delete a single value in a array**

Syntax: np.delete(array,indexnumber) 

```python
a  = np.array([15,19,24,37,42,26])
print(a)
```

```python
a3 = np.delete(a,1)
print(a3)
```

**Delete a multiple values in a array**

Syntax : np.delete(array,[multiple index numbers]]

```python
a4  = np.delete(a,[1,3,4])         # deleting multiple values
print(a4)
```

**Sorting an Array**

```python
a = np.array([4,6,8,2,1])

b = np.sort(a)
print(b)
```

**shallow copy**

```python
a= np.array([1,2,3])
b= a.copy()

b[1]=20
print(a)
print(b)
```
