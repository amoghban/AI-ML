# Modification/Data MAnipulation

**replace a single value with single value**

```python
s.iloc[2] = 32
print(s)
```

```python
s.loc["c"] = 72
print(s)
```

```python
s.iloc[[0,3,4]] =20
print(s)
```

**replace multiple values with single value**

```python
s.iloc[[0,3]] = 4
print(s)
```

**replace multiple values with multiple values**

```python
s.iloc[0:3] = [5,15,25]
print(s)
```

**add a new value into series**

```python
s.loc["f"] = 55
print(s)
```

```python
s.loc['i'], s.loc['j'] = 60,70
print(s)
```

```python
s
```

**remove a single value**

Syntax : s.drop(index=[list of index name(s)])

```python
s = s.drop(index=["d"])
print(s)
```

**remove multiple values**

```python
s = s.drop(index=["e","f"])
print(s)
```
